```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("SWYNEX - Task 2")
print("Exploratory Data Analysis")
print("Libraries imported successfully!")
```

```python

import os

print("Available files:")
for file in os.listdir():
    print(file)
```

```python

# Load cleaned dataset
df = pd.read_csv("cleaned_sales_data.csv")

# Display first 5 rows
print("Dataset Loaded Successfully!")
display(df.head())
```

```python

print("Dataset Shape:", df.shape)

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe(include="all"))

print("\nMissing Values:")
print(df.isnull().sum())
```

```python

# Numerical and categorical columns

numerical_columns = df.select_dtypes(include=np.number).columns.tolist()
categorical_columns = df.select_dtypes(include="object").columns.tolist()

print("Numerical Columns:")
print(numerical_columns)

print("\nCategorical Columns:")
print(categorical_columns)

print("\nDuplicate Rows:", df.duplicated().sum())

print("\nUnique Values:")
for col in df.columns:
    print(col, ":", df[col].nunique())
```

```python

categorical_columns = df.select_dtypes(
    include=["object", "str"]
).columns.tolist()

print("Categorical Columns:", categorical_columns)
```

```python

plt.figure(figsize=(10, 5))

if "Total" in df.columns:
    sns.histplot(df["Total"], bins=20, kde=True)
    plt.title("Sales Distribution Analysis")
    plt.xlabel("Total Sales")
    plt.ylabel("Frequency")
    plt.show()
else:
    print("Total column not found. Available columns:", df.columns.tolist())
```

```python

plt.figure(figsize=(12, 7))

sns.heatmap(
    df.select_dtypes(include=np.number).corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.show()
```

```python

# Category-wise Sales Analysis

category_sales = df.groupby("Product line")["Total"].sum().sort_values(ascending=False)

plt.figure(figsize=(12, 6))

category_sales.plot(kind="bar", edgecolor="black")

plt.title("Category-wise Total Sales Analysis")
plt.xlabel("Product Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=35)
plt.tight_layout()
plt.show()

print("Category-wise Sales:")
display(category_sales)
```

```python

branch_sales = df.groupby("Branch")["Total"].sum().sort_values(ascending=False)

plt.figure(figsize=(8, 5))

branch_sales.plot(kind="bar", edgecolor="black")

plt.title("Branch-wise Sales Performance")
plt.xlabel("Branch")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.show()

display(branch_sales)
```

```python

payment_count = df["Payment"].value_counts()

plt.figure(figsize=(7, 7))

plt.pie(
    payment_count,
    labels=payment_count.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Customer Payment Method Distribution")
plt.show()

display(payment_count)
```

```python

print("=" * 55)
print("SWYNEX - EXPLORATORY DATA ANALYSIS")
print("FIVE KEY BUSINESS INSIGHTS")
print("=" * 55)

# Insight 1: Highest selling product category
category_sales = df.groupby("Product line")["Total"].sum()
print("\n1. TOP PERFORMING PRODUCT CATEGORY:")
print(category_sales.idxmax(), "-", round(category_sales.max(), 2))

# Insight 2: Lowest selling product category
print("\n2. LOWEST PERFORMING PRODUCT CATEGORY:")
print(category_sales.idxmin(), "-", round(category_sales.min(), 2))

# Insight 3: Best performing branch
branch_sales = df.groupby("Branch")["Total"].sum()
print("\n3. BEST PERFORMING BRANCH:")
print(branch_sales.idxmax(), "-", round(branch_sales.max(), 2))

# Insight 4: Most popular payment method
payment = df["Payment"].value_counts()
print("\n4. MOST POPULAR PAYMENT METHOD:")
print(payment.idxmax(), "-", payment.max(), "transactions")

# Insight 5: Average customer rating
rating_column = next(
    (c for c in df.columns if c.lower() == "rating"),
    None
)

if rating_column:
    print("\n5. AVERAGE CUSTOMER RATING:")
    print(round(df[rating_column].mean(), 2))
```

```python

print("FINAL EDA SUMMARY")
print("-" * 40)

print("Total Records:", len(df))
print("Total Columns:", len(df.columns))
print("Total Sales:", round(df["Total"].sum(), 2))
print("Average Sales:", round(df["Total"].mean(), 2))
print("Maximum Sale:", round(df["Total"].max(), 2))
print("Minimum Sale:", round(df["Total"].min(), 2))
print("Duplicate Records:", df.duplicated().sum())
print("Missing Values:", df.isnull().sum().sum())

print("\nEDA Analysis Completed Successfully!")
```
